```python
# Find maximum between three numbers

def maximum(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    return c

a = int(input())
b = int(input())
c = int(input())

print(maximum(a, b, c))
```

```python
# Calculate the total electricity bill

def electricity_bill(units):
    if units <= 50:
        bill = units * 0.50
    elif units <= 150:
        bill = 50 * 0.50 + (units - 50) * 0.75
    elif units <= 250:
        bill = 50 * 0.50 + 100 * 0.75 + (units - 150) * 1.20
    else:
        bill = 50 * 0.50 + 100 * 0.75 + 100 * 1.20 + (units - 250) * 1.50

    return bill + bill * 0.20

units = int(input())

print(electricity_bill(units))
```

```python
# Write even, odd and prime numbers to separate files

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


with open("numbers.txt", "r") as file:
    numbers = list(map(int, file.read().split()))


with open("even.txt", "w") as even_file:
    for n in numbers:
        if n % 2 == 0:
            even_file.write(str(n) + "\n")


with open("odd.txt", "w") as odd_file:
    for n in numbers:
        if n % 2 != 0:
            odd_file.write(str(n) + "\n")


with open("prime.txt", "w") as prime_file:
    for n in numbers:
        if is_prime(n):
            prime_file.write(str(n) + "\n")

print("Files created successfully")
```

```python
# Display ATM transaction

def atm():
    balance = float(input("Enter initial balance: "))

    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")

    choice = int(input("Enter choice: "))

    if choice == 1:
        print("Balance:", balance)

    elif choice == 2:
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print("Updated balance:", balance)

    elif choice == 3:
        amount = float(input("Enter withdrawal amount: "))

        if amount <= balance:
            balance -= amount
            print("Updated balance:", balance)
        else:
            print("Insufficient balance")

    else:
        print("Invalid choice")


atm()
```
